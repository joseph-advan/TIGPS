# Data Exploration: TIGPS 2025 Student Data (Step 0)

**Objective:** 
To perform a comprehensive initial inspection of the 2025 dataset (`TIGPS_W3_student_convertedviacode_20260103_fulltext.csv`).

**Key Exploration Steps:**
1.  **Data Loading & Structure:** Check dimensions, columns, and data types.
2.  **Duplicate Check:** Identify duplicate headers or student IDs.
3.  **Missing Value Analysis:** Identify columns with high missingness.
4.  **Abnormal Value Detection:** Look for common placeholders (e.g., -9, -99, 999).
5.  **Categorical Content:** Preview string columns to understand response formats.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set display options
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

## 1. Load Data

```python
# Input Path
input_path = r"..\..\..\Data\2025data\TIGPS_W3_student_convertedviacode_20260103_fulltext.csv"

try:
    df = pd.read_csv(input_path)
    print(f"Data loaded successfully.")
    print(f"Shape: {df.shape}")
except FileNotFoundError:
    print(f"Error: File not found at {input_path}")
```

## 3. Duplicate Analysis

```python
# Check for fully duplicate rows
dup_rows = df.duplicated().sum()
print(f"Fully duplicate rows: {dup_rows}")

# Check for duplicate IDs
# Identify probable ID column (usually 'TIGPS ID' or similar based on previous tasks)
id_cols = [col for col in df.columns if 'ID' in col.upper()]
print(f"Potential ID columns found: {id_cols}")

for col in id_cols:
    dup_ids = df.duplicated(subset=[col]).sum()
    print(f"Duplicate entries in '{col}': {dup_ids}")
```

## 4. Missing Value Analysis

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100

# Filter to show only columns with any missing values
missing_report = pd.DataFrame({'Missing': missing, 'Pct': missing_pct})
missing_report = missing_report[missing_report['Missing'] > 0].sort_values(by='Pct', ascending=False)

print(f"Columns with missing values: {len(missing_report)}")
print("Top 20 Missing columns:")
display(missing_report)
```

## 5. Abnormal Value Detection (Placeholders)

Checking for common survey placeholders like -9, -99, or strings resembling them.

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns
print("Checking for negative values in numeric columns (Potential Map for NaN):")

placeholder_summary = []

for col in numeric_cols:
    neg_count = (df[col] < 0).sum()
    if neg_count > 0:
        uniques = df[df[col] < 0][col].unique()
        placeholder_summary.append({'Column': col, 'Negative_Count': neg_count, 'Values': uniques})

if placeholder_summary:
    place_df = pd.DataFrame(placeholder_summary).sort_values(by='Negative_Count', ascending=False)
    display(place_df.head(20))
else:
    print("No negative values found in numeric columns.")
```

## 6. Categorical Content Preview

Checking unique values in object columns to see if encoding is needed.

```python
obj_cols = df.select_dtypes(include=['object']).columns

print(f"Categorical Columns: {len(obj_cols)}")

for col in obj_cols[:10]: # Preview first 10 to avoid clutter
    uniques = df[col].dropna().unique()
    print(f"\nColumn: {col}")
    print(f"Unique Count: {len(uniques)}")
    if len(uniques) < 20:
        print(f"Values: {uniques}")
    else:
        print(f"Sample Values: {uniques[:5]}...")
```
