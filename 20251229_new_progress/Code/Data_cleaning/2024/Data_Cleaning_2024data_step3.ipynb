# Data Cleaning Step 3: W2 Mental Health Columns

**Objective:**
1. Load `TIGPS_W2_studentdata_ver3_encoded.csv`.
2. Remove rows where `v55_1` to `v55_14` have missing values (NaN).
3. Shift the scale of `v55_1` to `v55_14` from **0-4** to **1-5** (add 1 to all values).
4. Save the result to a new file.

```python
import pandas as pd
import numpy as np
import os
```

```python
# Define file paths
input_file = r"C:\Users\user\Desktop\TIGPS_PLAN_DATA\TIGPS\20251229_new_progress\Data\2024data\TIGPS_W2_studentdata_ver3_encoded.csv"
output_file = r"C:\Users\user\Desktop\TIGPS_PLAN_DATA\TIGPS\20251229_new_progress\Data\2024data\TIGPS_W2_studentdata_ver4_cleaned_mental.csv"

print(f"Input file: {input_file}")
print(f"Output file: {output_file}")
```

```python
# Load the dataset
if os.path.exists(input_file):
    df = pd.read_csv(input_file, low_memory=False)
    print(f"Original shape: {df.shape}")
else:
    print("Error: Input file not found.")
```

```python
# Identify target columns
target_cols = [f'v55_{i}' for i in range(1, 15)]
print(f"Target columns: {target_cols}")
```

```python
# Step 1: Remove rows with missing values in target columns

# Check missing before
missing_before = df[target_cols].isnull().sum()
print("Missing values before cleaning:\n", missing_before)

# Drop rows
df_cleaned = df.dropna(subset=target_cols).copy()
print(f"Shape after dropping NaNs: {df_cleaned.shape}")
print(f"Rows removed: {len(df) - len(df_cleaned)}")
```

```python
# Step 2: Shift value scale from 0-4 to 1-5 (Add 1)

print("Range before shift:")
print(f"Min: {df_cleaned[target_cols].min().min()}, Max: {df_cleaned[target_cols].max().max()}")

# Add 1 to all target columns
df_cleaned[target_cols] = df_cleaned[target_cols] + 1

print("Range after shift:")
print(f"Min: {df_cleaned[target_cols].min().min()}, Max: {df_cleaned[target_cols].max().max()}")
```

```python
# Save the new file
df_cleaned.to_csv(output_file, index=False)
print(f"Saved cleaned file to: {output_file}")
```
