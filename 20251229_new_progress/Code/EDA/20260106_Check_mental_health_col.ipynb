# Mental Health Question Columns Check

This notebook checks for missing values and anomalies in mental health related columns for Wave 2 and Wave 3 datasets.

**Targets:**
- **W2**: `v55_1` to `v55_14` in `C:\Users\user\Desktop\TIGPS_PLAN_DATA\TIGPS\20251229_new_progress\Data\2024data\TIGPS_W2_studentdata_ver3_encoded.csv`
- **W3**: `54-1` to `54-14` in `C:\Users\user\Desktop\TIGPS_PLAN_DATA\TIGPS\20251229_new_progress\Data\2025data\TIGPS_W3_student_studentdata_ver2_cleaned_cols.csv`

```python
import pandas as pd
import numpy as np
import os
```

```python
def check_columns(df, cols, dataset_name):
    print(f"--- Checking {dataset_name} ---")
    print(f"Total rows: {len(df)}")
    
    # Check if columns exist
    missing_cols = [c for c in cols if c not in df.columns]
    if missing_cols:
        print(f"WARNING: The following columns were not found in the dataframe: {missing_cols}")
        cols = [c for c in cols if c in df.columns]
    
    if not cols:
        print("No columns to check.")
        return

    # Missing values
    missing_counts = df[cols].isnull().sum()
    missing_pct = (missing_counts / len(df)) * 100
    
    print("\nMissing Values:")
    missing_df = pd.DataFrame({'Missing Count': missing_counts, 'Missing %': missing_pct})
    print(missing_df)
    
    print("\nValue Counts & Anomalies Check:")
    for col in cols:
        # Drop NaN for unique and checking anomalies
        valid_data = df[col].dropna()
        unique_vals = sorted(valid_data.unique())
        print(f"\nColumn: {col}")
        print(f"Unique values: {unique_vals}")
        # Check for values outside typical likert range
        print(df[col].value_counts().sort_index())
```

## Check W2 Data (`v55_1` - `v55_14`)

```python
w2_path = r"C:\Users\user\Desktop\TIGPS_PLAN_DATA\TIGPS\20251229_new_progress\Data\2024data\TIGPS_W2_studentdata_ver3_encoded.csv"

if os.path.exists(w2_path):
    try:
        # Attempt to read with default engine, fall back if needed
        df_w2 = pd.read_csv(w2_path, low_memory=False)
        
        w2_cols = [f'v55_{i}' for i in range(1, 15)]
        check_columns(df_w2, w2_cols, "Wave 2 (W2)")
        
    except Exception as e:
        print(f"Error reading W2 file: {e}")
else:
    print(f"File not found: {w2_path}")
```

## Check W3 Data (`54-1` - `54-14`)

```python
w3_path = r"C:\Users\user\Desktop\TIGPS_PLAN_DATA\TIGPS\20251229_new_progress\Data\2025data\TIGPS_W3_student_studentdata_ver2_cleaned_cols.csv"

if os.path.exists(w3_path):
    try:
        df_w3 = pd.read_csv(w3_path, low_memory=False)
        
        w3_cols = [f'54-{i}' for i in range(1, 15)]
        check_columns(df_w3, w3_cols, "Wave 3 (W3)")
        
    except Exception as e:
        print(f"Error reading W3 file: {e}")
else:
    print(f"File not found: {w3_path}")
```
